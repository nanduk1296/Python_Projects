Task 2-Data Cleaning

```python
#Handling Missing Values
#Numerical columns: replacing with median
num_cols = ['age', 'likes', 'comments', 'shares',
            'watch_time_sec', 'impression_count',
            'follower_count', 'engagement_rate']

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

#Categorical columns: replacing with mode
cat_cols = ['gender', 'country', 'post_type',
            'post_category', 'device_type',
            'sentiment', 'hashtags']

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
#Removing duplicates
df = df.drop_duplicates()

print("New Shape:", df.shape)
```

```python
# Data Type Coversion
df['posted_at'] = pd.to_datetime(df['posted_at'])

df['is_verified'] = df['is_verified'].astype(bool)
df.info()
```

```python
#Standardising Text Columns
text_cols = ['gender', 'country', 'post_type',
             'post_category', 'device_type',
             'sentiment']

for col in text_cols:
    df[col] = df[col].str.strip().str.lower()
```

```python
#Checking for Unrealistic Values
df[['likes', 'comments', 'shares']].describe()

plt.figure(figsize=(12,5))

plt.subplot(1,3,1)
sns.boxplot(y=df['likes'])
plt.title('Likes')

plt.subplot(1,3,2)
sns.boxplot(y=df['comments'])
plt.title('Comments')

plt.subplot(1,3,3)
sns.boxplot(y=df['shares'])
plt.title('Shares')

plt.tight_layout()
plt.show()

#Instead of deleting rows, extreme values will be capped at the upper and lower bounds
def cap_outliers(column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df[column] = np.where(
        df[column] > upper_bound,
        upper_bound,
        np.where(
            df[column] < lower_bound,
            lower_bound,
            df[column]
        )
    )

for col in ['likes', 'comments', 'shares']:
    cap_outliers(col)
#Recheck
df[['likes', 'comments', 'shares']].describe()
```

#Executive Summary
###Project Overview
This project analyzed a social media engagement dataset containing user demographics, content characteristics, engagement metrics, sentiment data, and platform usage information. The objective was to identify the key factors influencing engagement and derive actionable insights from user behavior and content performance.

###Content Performance Analysis
The analysis revealed that video posts generated the highest average engagement rate (1.122), making them the most effective content format. Text-based posts ranked second, while reels recorded the lowest engagement.

Among content categories, food-related content was the strongest performer with an average engagement rate of 1.359, followed by technology and lifestyle content.

Brazil achieved the highest engagement rate among all countries analyzed, followed by Australia and France.

###User Trends Analysis
User engagement varied across age groups. The 45-60 age group demonstrated the highest engagement rate, followed by users aged 18-25 years.

Verified accounts outperformed non-verified accounts, achieving an average engagement rate of 1.054 compared to 0.955 for non-verified accounts.

###Behavioral Insights
Mobile users recorded the highest average watch time, indicating that mobile devices remain the dominant medium for content consumption.

The dataset contained only posting dates and did not include posting timestamps. Therefore, a reliable analysis of the best time of day for impressions could not be performed.

###Sentiment Analysis
Negative sentiment posts achieved the highest average engagement rate, followed by neutral sentiment posts. Positive sentiment posts generated the lowest engagement rate.

Negative posts also recorded the highest average likes and shares, while neutral posts generated the highest average comment activity.

###Conclusion
The findings suggest that video content, food-related topics, mobile-first strategies, and verified accounts are key drivers of engagement. Additionally, audience demographics and sentiment significantly influence content performance.

###Key Findings
Video posts generated the highest engagement.
Food was the top-performing content category.
Brazil recorded the highest average engagement rate.
Users aged 45-60 were the most engaged audience segment.
Verified accounts outperformed non-verified accounts.
Mobile devices generated the highest watch time.
Negative sentiment posts drove the highest engagement.
Best-time-of-day analysis was not possible due to missing timestamp data.
